```python
data_dir='/home/vagrant/jupyter/StaffScratch/diaw/Ionboost/'
folder ='test'
```

```python
cols =['x0','xt','v','phi','E','ni','ne','rho','nhot','ncold','nss','difnor','dWhot','SWhot','charge']
fname = path.join(data_dir,folder,'profil.txt')
df=read_csv(fname,comment="#",header=None,sep='\s+',names=cols,index_col=False,skiprows=4)
```

```python
fig,ax1 = plt.subplots(2,2,figsize = (12,6))
plt.matplotlib.rc('xtick', labelsize=15)
plt.matplotlib.rc('ytick', labelsize=15)
LW=2

ax1[0][0].plot(df['xt'],df['ni'], linewidth=LW)
ax1[0][0].plot(df['xt'],df['ne'], linewidth=LW)
ax1[0][1].plot(df['xt'],df['rho'], linewidth=LW)
ax1[1][0].plot(df['xt'],df['v'], linewidth=LW)
ax1[1][1].plot(df['xt'],df['E'], linewidth=LW)

ax1[0][0].set_xlim(-100,400)
ax1[0][1].set_xlim(-100,400)
ax1[1][0].set_xlim(-10,400)
ax1[1][1].set_xlim(-100,400)

ax1[0][0].set_yscale('log')
ax1[0][0].set_ylim(1.e-4,1.2)
ax1[1][0].set_xlabel(r'x [$\lambda_{Dh}$]', size=15)
ax1[1][1].set_xlabel(r'x [$\lambda_{Dh}$]', size=15)
ax1[0][0].set_ylabel(r'$n_i$', size=15)
ax1[0][1].set_ylabel(r'$n_e-Z_in_i$', size=15)
ax1[1][0].set_ylabel(r'v', size=15)
ax1[1][1].set_ylabel(r'E', size=15)
ax1[0][0].grid(alpha=0.3)
ax1[0][1].grid(alpha=0.3)
ax1[1][1].grid(alpha=0.3)
ax1[1][0].grid(alpha=0.3)
fig.subplots_adjust(hspace=0.3, wspace=0.4)
plt.savefig(folder+'spatial_profils.png', bbox_inches = 'tight')     
```
