## Fluid simulations of a plasma expansion into a vacuum.
#### This test case reproduces some of the results of Mora's 2003 PRL paper.

```python
from os import path
from pandas import read_csv
import matplotlib.pyplot as plt
import matplotlib as mpl
```

```python
%matplotlib inline
```

```python
data_dir='/home/vagrant/jupyter/StaffScratch/diaw/Ionboost/'
data_dir='/Users/diaw/Ionboost/'
folder ='test'
```

```python
cols =['x0','xt','v','phi','E','ni','ne','rho','nhot','ncold','nss','difnor','dWhot','SWhot','charge']
fname = path.join(data_dir,folder,'profil.txt')
df=read_csv(fname,comment="#",header=None,sep='\s+',names=cols,index_col=False,skiprows=4)
```

## Let us now look at the charge separation

```python
wt=50
with mpl.style.context('classic'):
    mpl.rc('font', size = 18)
    mpl.rcParams['figure.figsize'] = 6,4
    fig,ax1 = plt.subplots(1,1)
    ax1.plot(df['xt']/wt,df['rho'], linewidth=2)
    ax1.set_xlim(-2,10)
    ax1.set_xlabel(r'$x/c_s t$', size=15)
    ax1.set_ylabel(r'$(n_e-Z_in_i)/n_{e0}$', size=15)
#     ax1.grid(alpha=0.3)
    ax1.set_title('Charge separation at time $\omega_{pi}t=50$', fontsize=16)
    ax1.grid(linestyle='dashed', linewidth='0.5')
    plt.tight_layout()

    fig.subplots_adjust(hspace=0.3, wspace=0.4)
    plt.savefig('charge_separation.png', bbox_inches = 'tight')     
    plt.show()
```

Note that the ion front stands at  $x/c_st \approx 5.59$

## Electric field

```python
wt=50
with mpl.style.context('classic'):
    mpl.rc('font', size = 18)
    mpl.rcParams['figure.figsize'] = 6,4
    fig,ax1 = plt.subplots(1,1)
    ax1.plot(df['xt']/wt,df['E'], linewidth=2)
    ax1.set_xlim(-2,10)
    ax1.set_xlabel(r'$x/c_s t$', size=15)
    ax1.set_ylabel(r'$E/E_0$', size=15)
#     ax1.grid(alpha=0.3)
    ax1.set_title('Electric field at time $\omega_{pi}t=50$', fontsize=16)
    ax1.grid(linestyle='dashed', linewidth='0.5')
    plt.tight_layout()

    fig.subplots_adjust(hspace=0.3, wspace=0.4)
    plt.savefig('electric_field.png', bbox_inches = 'tight')     
    plt.show()
```

The ion front stands at  $x/c_st \approx 5.59$.

Note that the electric field at the ion front is approximately twice the self-similar field.

## Density profiles

```python
wt=50
with mpl.style.context('rs_paper'):
    mpl.rc('font', size = 18)
    mpl.rcParams['figure.figsize'] = 6,4
    fig,ax1 = plt.subplots(1,1)
    ax1.plot(df['xt']/wt,df['ne'], label='$n_e$', linewidth=2)
    ax1.plot(df['xt']/wt,df['ni'], label='$n_i$', linewidth=2)
    ax1.set_xlim(-2,10)
    ax1.set_xlabel(r'$x/c_s t$', size=15)
    ax1.set_ylabel(r'$n_e/n_{e0},Z_in_i/n_{e0}$', size=15)
    ax1.set_title('Structure of the ion front at time $\omega_{pi}t=50$', fontsize=16)
    ax1.grid(linestyle='dashed', linewidth='0.5')
    ax1.set_yscale('log')
    ax1.set_ylim(1.e-4,10)
    ax1.legend(loc='best')
    plt.tight_layout()

    plt.savefig('densities_profile.png', bbox_inches = 'tight')     
    plt.show()
```

## Energy spectrum

```python
cols =['vmoy', 'Emoy', 'dndv','dndE']
fname = path.join(data_dir,folder,'spectres.txt')
df=read_csv(fname,comment="#",header=None,sep='\s+',names=cols,index_col=False,skiprows=1)
```

```python
wt=50
with mpl.style.context('rs_paper'):
    mpl.rc('font', size = 18)
    mpl.rcParams['figure.figsize'] = 6,4
    fig,ax1 = plt.subplots(1,1)
    ax1.plot(df['Emoy'],df['dndE'], label='$n_e$', linewidth=2)
#     ax1.set_xlim(-2,10)
    ax1.set_xlabel(r'$E$', size=15)
    ax1.set_ylabel(r'$dN/DE}$', size=15)
    ax1.set_title('Energy spectrum per unit surface at time $\omega_{pi}t=50$', fontsize=16)
    ax1.grid(linestyle='dashed', linewidth='0.5')
    ax1.set_yscale('log')
#     ax1.set_ylim(1.e-4,10)
    ax1.legend(loc='best')
    plt.tight_layout()

    plt.savefig('energy_spectrum.png', bbox_inches = 'tight')     
    plt.show()
```

Energy is normalized to  $Zk_BT_e$, and the number of ions per unit surface 
and unit energy is normalized to $n_{i0} \lambda_{D0}/Zk_BT_e$.
