```python
data_dir='/Users/42d/Ionboost/'
folder ='scripts'
```

```python
cols =['x0','xt','v','phi','E','ni','ne','rho','nhot','ncold','nss','difnor','dWhot','SWhot','charge']
fname = path.join(data_dir,folder,'profil.txt')
df=read_csv(fname,comment="#",header=None,sep='\s+',names=cols,index_col=False,skiprows=0)

 # x0test xttest vtest itest E
# cols =['x0test', 'xttest', 'vtest', 'itest', 'E']
# fname_ = path.join(data_dir,folder,'profil_test')
# df_=read_csv(fname_,comment="#",header=None,sep='\s+',names=cols,index_col=False,skiprows=0)
```

## Let us now look at the charge separation

Note that the ion front stands at  $x/c_st \approx 5.59$

## Potential

```python
fig,ax1 = plt.subplots(1,1)
wt=50
E0=30.972912245662215/1.e3
lam_e=0.00010681297986043954
ax1.plot(df['xt']*lam_e,df['E']*E0, linewidth=2)

ax1.set_xlim(1.e-4,.2)
# ax1.set_ylim(1.e-4,1.2)
ax1.set_xlabel(r'$x/c_s t$', size=15)
ax1.set_ylabel(r'$E/E_0$', size=15)
# ax1.set_yscale('log')
# ax1.set_xscale('log')
#     ax1.grid(alpha=0.3)
ax1.set_title('Electric field at time $\omega_{pi}t=50$', fontsize=16)
ax1.grid(linestyle='dashed', linewidth='0.5')
plt.tight_layout()

fig.subplots_adjust(hspace=0.3, wspace=0.4)
plt.savefig('electric_field.png', bbox_inches = 'tight')     
plt.show()
```

The ion front stands at  $x/c_st \approx 5.59$.

Note that the electric field at the ion front is approximately twice the self-similar field.

## Density profiles

```python
fig,ax1 = plt.subplots(1,1)
time=1
wt=1
ax1.plot(df['xt']/wt/time,df['ne'], label='$n_e$', linewidth=2)
ax1.plot(df['xt']/wt/time,df['ni'], label='$n_i$', linewidth=2)
ax1.plot(df['xt']/wt/time,df['nss'], label='$n_ss$', linewidth=2)
ax1.set_xlim(-200,200)
ax1.set_xlabel(r'$x/c_s t$', size=15)
ax1.set_ylabel(r'$n_e/n_{e0},Z_in_i/n_{e0}$', size=15)
ax1.set_title('Structure of the ion front at time $\omega_{pi}t=50$', fontsize=16)
ax1.grid(linestyle='dashed', linewidth='0.5')
# ax1.set_yscale('log')
# ax1.set_ylim(1.e-4,10)
ax1.legend(loc='best')
plt.tight_layout()

plt.savefig('densities_profile.png', bbox_inches = 'tight')     
plt.show()
```

Energy is normalized to  $Zk_BT_e$, and the number of ions per unit surface 
and unit energy is normalized to $n_{i0} \lambda_{D0}/Zk_BT_e$.
